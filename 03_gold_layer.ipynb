# 03 Gold Layer — Aggregations & Feature Engineering

The Gold layer creates:
1. Business-level aggregated metrics
2. Machine learning features

This layer is used for analytics and ML models.

```python
SILVER_TABLE = "silver_spotify"

df = spark.table(SILVER_TABLE)

display(df)
```

## Business Aggregations (Genre-level insights)

```python
gold_genre_df = spark.sql("""
SELECT
    track_genre,
    COUNT(*) AS total_tracks,
    AVG(popularity) AS avg_popularity,
    AVG(danceability) AS avg_danceability,
    AVG(energy) AS avg_energy,
    AVG(valence) AS avg_valence,
    AVG(tempo) AS avg_tempo
FROM silver_spotify
GROUP BY track_genre
ORDER BY avg_popularity DESC
""")

display(gold_genre_df)
```

```python
gold_genre_df.write.format("delta").mode("overwrite").saveAsTable("gold_genre_metrics")

print("Gold aggregation table created: gold_genre_metrics")
```

## Feature Engineering for Machine Learning

We create a target variable `popularity_tier`:
- High: popularity >= 70
- Medium: 40–69
- Low: < 40

```python
from pyspark.sql.functions import when, col

gold_ml_df = df.withColumn(
    "popularity_tier",
    when(col("popularity") >= 70, "high")
    .when(col("popularity") >= 40, "medium")
    .otherwise("low")
)

display(gold_ml_df)
```

```python
gold_ml_df.write.format("delta").mode("overwrite").saveAsTable("gold_spotify_ml_features")

print("Gold ML table created: gold_spotify_ml_features")
```
