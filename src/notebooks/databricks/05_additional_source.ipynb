# Task 2: Additional Data Source Integration

Instead of using an external dataset, we create a derived dataset from Spotify data:

- Artist-level aggregation (acts as second source)
- Join back with track-level data
- Generate new insights

This simulates real-world data integration.

```python
df = spark.table("silver_spotify")

display(df)
```

```python
artist_df = spark.sql("""
SELECT
    artists,
    COUNT(*) AS total_tracks,
    AVG(popularity) AS avg_artist_popularity,
    AVG(energy) AS avg_energy
FROM silver_spotify
GROUP BY artists
""")

display(artist_df)
```

```python
combined_df = df.join(artist_df, on="artists", how="left")

display(combined_df)
```

```python
# Top artists by popularity
top_artists = spark.sql("""
SELECT artists, AVG(popularity) AS avg_popularity
FROM silver_spotify
GROUP BY artists
ORDER BY avg_popularity DESC
LIMIT 10
""")

display(top_artists)
```

```python
combined_df.select("popularity", "avg_artist_popularity").describe().show()
```

```python
combined_df.write.format("delta").mode("overwrite").saveAsTable("gold_combined_dataset")

print("Combined dataset created")
```
